```python
import  pandas as pd
df = pd.read_excel('/content/08_Electric Vehicle (4).xlsx')
```

```python
df.isnull().sum()
```

```python
df['City'] = df['City'].fillna(df['City'].mode()[0])
```

```python
df['Model Year'] = df['Model Year'].fillna(df['Model Year'].median())
```

```python
df['Make'] = df['Make'].fillna(df['Make'].mode()[0])
```

```python
df['Model'] = df['Model'].fillna(df['Model'].mode()[0])
```

```python
df['Vehicle Identification Number'] = df['Vehicle Identification Number'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
df['model'] = df['model'].astype(str)
df.loc[df['model'].str.strip().str.fullmatch(r'\d+'), 'model'] = 'Unknown'
```

```python
df.to_excel('08_Electric Vehicle.xlsx')
```

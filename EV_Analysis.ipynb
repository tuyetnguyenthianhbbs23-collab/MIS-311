```python
import  pandas as pd
df = pd.read_excel('/content/08_Electric Vehicle (1).xlsx')
```

```python
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['city'] = df['city'].fillna(df['city'].mode()[0])
```

```python
df['model_year'] = df['model_year'].fillna(df['model_year'].median())
```

```python
df['make'] = df['make'].fillna(df['make'].mode()[0])
```

```python
df['model'] = df['model'].fillna(df['model'].mode()[0])
```

```python
df['vehicle_identification_number'] = df['vehicle_identification_number'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
df['model'] = df['model'].astype(str)
df.loc[df['model'].str.strip().str.fullmatch(r'\d+'), 'model'] = 'Unknown'
```

```python
df.describe(include='all')
```

```python
df.to_excel('08_Electric Vehicle.xlsx')
```
